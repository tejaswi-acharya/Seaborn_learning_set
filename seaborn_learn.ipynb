```python
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
sns.get_dataset_names()
```

```python
penguins=sns.load_dataset('penguins')
penguins.head()
```

```python
penguins['species'].value_counts()
```

```python
penguins['island'].value_counts()
```

```python

sns.set_style("ticks")  #white,whitegrid,darkgrid,ticks
sns.scatterplot(data=penguins, x="flipper_length_mm", y="body_mass_g", hue= "sex",palette="deep")
sns.despine(left=True)

#set_context()  paper,notebook,talk,poster  used to override

sns.set_context("paper")



#palette  decides what colors are assigned to different categories, how many colors to generate , style mood is pastel, bright or dark

#used in sns.scatterplot(...,palette="virdis")
#sns.barplot(...,palette="pastel")
#sns.countplot(...,palette="deep")
#sns.lineplot(...,palette="muted") flare,husl,tab10,Set2...for categories kind 
# Blues, Greens,OrRd, mako,rocket....for ordered data, like temperature, marks, values
# coolwarm,seismic,vlag,icefire....diverging palettes, used when data is centered around amidpoint like correlation

```

```python
sns.set_style("ticks")  #white,whitegrid,darkgrid,ticks
sns.scatterplot(data=penguins, x="flipper_length_mm", y="body_mass_g", hue= "island",palette="Set1")
sns.despine(left=True)

#Set1=bright and high contrast colors
#Set2=socter more pastel like colors
#Set3=a mix of soft and bold varied hues

sns.set_context("paper")
```

```python
#style parameter controls the shape of points or symbols of points
sns.scatterplot(data=penguins, x="species", y="body_mass_g", hue= "island",style="sex",palette="Dark2",alpha = 0.5)  #hue different color or each island, style="sex"-different marker shape for each species
sns.set_context("poster")
```

```python
sns.stripplot(data=penguins,x="species",y="body_mass_g", hue="island",dodge=True,jitter=0.2)
```

```python
#swarm plot-it is a categorical scatter plot in seaborn that arranges points very carefully so they never overlap givinb a clear view of the distribution of data
sns.swarmplot(data=penguins,x="species",y="body_mass_g", hue="island")
```

```python
sns.histplot(data=penguins,x="body_mass_g", hue="sex",multiple="stack")
sns.set_context("paper")
```

```python
#regression plot
sns.regplot(data=penguins, x="flipper_length_mm", y="body_mass_g",scatter=False)
```

```python
sns.set_style("darkgrid")
sns.lineplot(data=penguins,x="body_mass_g",y="flipper_length_mm",hue="sex",style="island")
```

```python
sns.jointplot(data=penguins,x="body_mass_g",y="flipper_length_mm",hue="sex", kind="kde")
```
